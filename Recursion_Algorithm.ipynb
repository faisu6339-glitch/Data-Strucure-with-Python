#**Basic Recursion**

##### **Explanation**
Basic recursion occurs when a computer function invokes itself as a subroutine to solve a smaller piece of a larger puzzle. A recursive function acts as an engine that continuously shrinks the input data, passing the reduced data into a fresh instance of itself. This downward cascade continues until the input shrinks to a universally recognized, trivial state that requires no further calculation.

To prevent infinite cascading, every valid recursive function mandates two distinct structural pillars. The first pillar is the base case, which acts as the definitive stopping condition. The second pillar is the recursive step, which performs a marginal calculation and invokes the function again with modified parameters. Without the base case, the function will call itself forever, leading to a catastrophic memory failure known as a stack overflow.

The primary purpose of basic recursion is to express complex algorithms in a clean, mathematical, and readable format. Problems that involve natural hierarchies, such as navigating a computer file system or generating mathematical sequences, map perfectly to recursive logic because the structure of the solution exactly matches the structure of the problem.

##### **Understanding Concept**
Imagine standing at the back of a very long line of people and needing to know exactly how many people are in the line. Instead of leaving your spot to walk forward and count every person, you tap the person directly in front of you and ask, "How many people are in front of you?" That person does not know, so they tap the person in front of them and ask the exact same question. This question propagates all the way to the very first person in the line. The first person answers, "Zero." They turn around and tell the second person, who adds one to the answer and passes it back. This backward chain reaction continues until the final answer reaches you at the back of the line.

##### **How it Works**
Step One: The main program invokes the recursive function, passing an initial set of data parameters.
Step Two: The function immediately evaluates the parameters against the defined base case condition.
Step Three: If the base case condition is met, the function halts further execution and returns a concrete, pre-defined value.
Step Four: If the base case is not met, the function performs a minor mathematical or logical operation to reduce the size or complexity of the input data.
Step Five: The function invokes itself again, passing the newly reduced data as the updated parameters.
Step Six: The function waits for the newly invoked instance to return a value, combines that returned value with its own minor operation, and passes the total result backward.
Edge Cases: A robust recursive function must account for invalid inputs, such as negative numbers in a factorial calculation, which would completely bypass a base case designed for zero and trigger an infinite loop.

#FACTORIAL

# Using Recursion

```python
def calculate_factorial(n):
  #base case condition
  if n < 0:
    return -1

  if n==0:
    return 1

  return n* calculate_factorial(n-1)
calculate_factorial(5)
```

# Using Loop

```python
def factorial(n):
    result = 1

    for i in range(1, n + 1):
        result *= i

    return result

print(factorial(5))
```

# FIBONACCI SERIES

# Using Recursion

```python
def fibonacci_series(n):
  if n < 0:
    return -1
  if n <=1:
    return n

  return fibonacci_series(n-1) + fibonacci_series(n-2)

print(fibonacci_series(6))
```

# Using Loop

```python
def fib(n):
    a = 0
    b = 1

    for _ in range(n):
        a, b = b, a + b

    return a
print(fib(6))
```

# SUM OF DIGITS

# Using Recursion

```python
def sum_of_digits(n):
    if n == 0:              # Base Case
        return 0

    return (n % 10) + sum_of_digits(n // 10)

print(sum_of_digits(1234))
```

# Using Loop

```python
def sum_of_digits(n):
    total = 0

    while n > 0:
        total += n % 10
        n //= 10

    return total

print(sum_of_digits(1234))
```

```python
def merge_sort(arr):

    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])

    right = merge_sort(arr[mid:])

    result = []

    i = 0
    j = 0

    while i < len(left) and j < len(right):

        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


arr = [8,3,5,4,7,6,1,2]

print(merge_sort(arr))
```

# Quick Sort

```python
def partition(arr, low, high):

    pivot = arr[high]

    i = low - 1

    for j in range(low, high):

        if arr[j] < pivot:

            i += 1

            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1


def quick_sort(arr, low, high):

    if low < high:

        pi = partition(arr, low, high)

        quick_sort(arr, low, pi - 1)

        quick_sort(arr, pi + 1, high)


arr = [8,3,5,4,7,6,1,2]

quick_sort(arr, 0, len(arr)-1)

print(arr)
```

**Quick Sort Recursion Tree:**

                    Pivot = 2

               /                 \

            [1]            [8,3,5,4,7,6]

                           Pivot = 6

               [3,5,4]               [8,7]

               Pivot=4

             /        \

           [3]        [5]

```python
def binary_search(arr, low, high, target):

    # Base Case
    if low > high:
        return -1

    mid = (low + high) // 2

    if arr[mid] == target:
        return mid

    elif target < arr[mid]:
        return binary_search(arr, low, mid - 1, target)

    else:
        return binary_search(arr, mid + 1, high, target)


# Example
arr = [5, 8, 12, 16, 23, 38, 45, 56]

target = 23

result = binary_search(arr, 0, len(arr) - 1, target)

if result != -1:
    print("Element found at index:", result)
else:
    print("Element not found")
```

```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid

        elif target < arr[mid]:
            high = mid - 1

        else:
            low = mid + 1

    return -1


# Example
arr = [5, 8, 12, 16, 23, 38, 45, 56]
target = 23

result = binary_search(arr, target)

if result != -1:
    print("Element found at index:", result)
else:
    print("Element not found")
```
