```python
class Node:
  def __init__(self, value):
    self.value = value
    self.next = None

a=Node(5)
b=Node(7)
c=Node(9)

a.next=b
b.next=c
c.next=None

print(a.next.value)
print(b.next.value)
print(c.next)
```

#Creating Head Node

```python
class Node:
  def __init__(self, value):
    self.value = value
    self.next = None

a=Node(5)
b=Node(7)
c=Node(9)

head=a
print(head.value)
a.next=b
print(head.next.value)
b.next=c
print(head.next.next.value)
```

#Using Traversing

```python
# Function to print linked list
def print_list(head):
    curr = head

    while curr is not None:
        print(curr.value, end=" ")
        curr = curr.next


print_list(head)
```

# Insertion

**Insertion at beg**

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Create nodes
a = Node(72)
b = Node(87)
c = Node(96)

# Link the nodes
a.next = b
b.next = c

# Head node
head = a
new_node = Node(10)
new_node.next = head
head = new_node


# Function to print linked list
def print_list(head):
    curr = head

    while curr is not None:
        print(curr.data, end=" ")
        curr = curr.next

    print()   # Move to next line after printing all nodes


print_list(head)
```
