# Stacks in Data Structures

A **stack** is a linear data structure that follows a particular order in which operations are performed. The order is **LIFO (Last In, First Out)**. Think of a stack of plates: you can only add a new plate to the top, and you can only remove the topmost plate.

### Key Operations:
*   **Push**: Adds an item to the top of the stack.
*   **Pop**: Removes the item from the top of the stack. It typically returns the removed item.
*   **Peek (or Top)**: Returns the top item without removing it.
*   **isEmpty**: Checks if the stack is empty.
*   **Size**: Returns the number of items in the stack.

### Stack Implementation in Python
Python's built-in `list` data type can be easily used to implement a stack. `append()` can be used for the push operation, and `pop()` for the pop operation.

```python
# Implementing a Stack using Python's list
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)
        print(f"Pushed: {item}. Stack now: {self.items}")

    def pop(self):
        if self.is_empty():
            return "Stack is empty"
        item = self.items.pop()
        print(f"Popped: {item}. Stack now: {self.items}")
        return item

    def peek(self):
        if self.is_empty():
            return "Stack is empty"
        item = self.items[-1]
        print(f"Peeked: {item}. (Top item without removing)")
        return item

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)

    def __str__(self):
        return str(self.items)

# Demonstration of Stack operations
my_stack = Stack()

print(f"Is stack empty? {my_stack.is_empty()}") # True
my_stack.push(10)
my_stack.push(20)
my_stack.push(30)

print(f"Current stack size: {my_stack.size()}") # 3
my_stack.peek() # 30

my_stack.pop() # 30
my_stack.pop() # 20

print(f"Is stack empty? {my_stack.is_empty()}") # False
my_stack.push(40)
print(f"Final stack: {my_stack}") # [10, 40]

my_stack.pop() # 40
my_stack.pop() # 10
my_stack.pop() # Stack is empty
```

# Implementing Stack through Array

```python
class Stack:
    def __init__(self):
        self.stack = []

    # Push element
    def push(self, item):
        self.stack.append(item)

    # Pop element
    def pop(self):
        if not self.is_empty():
            return self.stack.pop()
        else:
            return None

    # Check if stack is empty
    def is_empty(self):
        return len(self.stack) == 0

    # Peek / Top element
    def peek(self):
        if not self.is_empty():
            return self.stack[-1]
        else:
            return None

    # Get size of stack
    def size(self):
        return len(self.stack)

    # Display stack
    def display(self):
        if self.is_empty():
            print("Stack is empty")
        else:
            print("Stack:", self.stack)


# Create stack
s = Stack()

# Push elements
s.push(10)
s.push(20)
s.push(30)
s.push(40)

# Display stack
s.display()

# Peek
print("Top element:", s.peek())

# Size
print("Stack size:", s.size())

# Pop
print("Popped:", s.pop())
print("Popped:", s.pop())

# Display after pop
s.display()

# Peek again
print("Top element:", s.peek())

# Check empty
print("Is stack empty?", s.is_empty())
```

# Implementing Stack using LinkedList

```python
# Node class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Stack using Linked List
class Stack:
    def __init__(self):
        self.top = None

    # Push element
    def push(self, data):
        new_node = Node(data)

        # New node points to current top
        new_node.next = self.top

        # New node becomes top
        self.top = new_node

        print(data, "pushed into stack")

    # Pop element
    def pop(self):
        if self.is_empty():
            print("Stack Underflow")
            return None

        popped = self.top.data

        # Move top to next node
        self.top = self.top.next

        print(popped, "popped from stack")
        return popped

    # Peek / Top element
    def peek(self):
        if self.is_empty():
            print("Stack is empty")
            return None

        return self.top.data

    # Check if stack is empty
    def is_empty(self):
        return self.top is None

    # Display stack
    def display(self):
        if self.is_empty():
            print("Stack is empty")
            return

        curr = self.top

        print("Stack:")

        while curr is not None:
            print(curr.data)
            curr = curr.next


# Create stack
s = Stack()

# Push elements
s.push(10)
s.push(20)
s.push(30)
s.push(40)

# Display stack
s.display()

# Peek
print("Top element:", s.peek())

# Pop
s.pop()
s.pop()

# Display after pop
s.display()

# Peek again
print("Top element:", s.peek())

# Check empty
print("Is stack empty?", s.is_empty())
```

# Queues in Data Structures

A **queue** is a linear data structure that follows a particular order in which operations are performed. The order is **FIFO (First In, First Out)**. Think of a queue of people waiting to buy tickets: the first person in line is the first person to be served, and new people join the back of the line.

### Key Operations:
*   **Enqueue**: Adds an item to the rear (end) of the queue.
*   **Dequeue**: Removes the item from the front of the queue. It typically returns the removed item.
*   **Front (or Peek)**: Returns the front item without removing it.
*   **Rear**: Returns the rear item without removing it.
*   **isEmpty**: Checks if the queue is empty.
*   **Size**: Returns the number of items in the queue.

### Queue Implementation in Python
Python's built-in `list` can also be used to implement a queue, though for performance-critical applications, `collections.deque` is often preferred because `list.pop(0)` is less efficient than `deque.popleft()` as it requires shifting all subsequent elements.

```python
# Implementing a Queue using Python's list
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item) # Add to the end of the list
        print(f"Enqueued: {item}. Queue now: {self.items}")

    def dequeue(self):
        if self.is_empty():
            return "Queue is empty"
        item = self.items.pop(0) # Remove from the beginning of the list
        print(f"Dequeued: {item}. Queue now: {self.items}")
        return item

    def front(self):
        if self.is_empty():
            return "Queue is empty"
        item = self.items[0]
        print(f"Front: {item}. (Front item without removing)")
        return item

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)

    def __str__(self):
        return str(self.items)

# Demonstration of Queue operations
my_queue = Queue()

print(f"Is queue empty? {my_queue.is_empty()}") # True
my_queue.enqueue('A')
my_queue.enqueue('B')
my_queue.enqueue('C')

print(f"Current queue size: {my_queue.size()}") # 3
my_queue.front() # A

my_queue.dequeue() # A
my_queue.dequeue() # B

print(f"Is queue empty? {my_queue.is_empty()}") # False
my_queue.enqueue('D')
print(f"Final queue: {my_queue}") # ['C', 'D']

my_queue.dequeue() # C
my_queue.dequeue() # D
my_queue.dequeue() # Queue is empty
```

### Using `collections.deque` for more efficient Queue implementation
For more efficient queue operations, especially when dealing with large queues, Python's `collections.deque` (double-ended queue) is recommended as `append` and `popleft` are O(1) operations.

```python
from collections import deque

class EfficientQueue:
    def __init__(self):
        self.items = deque()

    def enqueue(self, item):
        self.items.append(item) # Add to the right end
        print(f"Enqueued: {item}. Queue now: {list(self.items)}")

    def dequeue(self):
        if self.is_empty():
            return "Queue is empty"
        item = self.items.popleft() # Remove from the left end
        print(f"Dequeued: {item}. Queue now: {list(self.items)}")
        return item

    def front(self):
        if self.is_empty():
            return "Queue is empty"
        item = self.items[0]
        print(f"Front: {item}. (Front item without removing)")
        return item

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)

    def __str__(self):
        return str(list(self.items))

# Demonstration of EfficientQueue operations
my_efficient_queue = EfficientQueue()

print(f"Is efficient queue empty? {my_efficient_queue.is_empty()}") # True
my_efficient_queue.enqueue('X')
my_efficient_queue.enqueue('Y')
my_efficient_queue.enqueue('Z')

print(f"Current efficient queue size: {my_efficient_queue.size()}") # 3
my_efficient_queue.front() # X

my_efficient_queue.dequeue() # X
my_efficient_queue.dequeue() # Y

print(f"Is efficient queue empty? {my_efficient_queue.is_empty()}") # False
my_efficient_queue.enqueue('W')
print(f"Final efficient queue: {my_efficient_queue}") # ['Z', 'W']

my_efficient_queue.dequeue() # Z
my_efficient_queue.dequeue() # W
my_efficient_queue.dequeue() # Queue is empty
```

# Implementing Queue using Linkedlist

```python
# Node class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Queue using Linked List
class Queue:
    def __init__(self):
        self.front = None
        self.rear = None

    # Enqueue - Insert element
    def enqueue(self, data):
        new_node = Node(data)

        # If queue is empty
        if self.rear is None:
            self.front = new_node
            self.rear = new_node
            return

        # Add new node at the end
        self.rear.next = new_node
        self.rear = new_node

    # Dequeue - Remove element
    def dequeue(self):
        # If queue is empty
        if self.front is None:
            print("Queue is empty")
            return None

        # Store front value
        removed = self.front.data

        # Move front to next node
        self.front = self.front.next

        # If queue becomes empty
        if self.front is None:
            self.rear = None

        return removed

    # Peek - Get front element
    def peek(self):
        if self.front is None:
            print("Queue is empty")
            return None

        return self.front.data

    # Check if queue is empty
    def is_empty(self):
        return self.front is None

    # Display queue
    def display(self):
        if self.front is None:
            print("Queue is empty")
            return

        curr = self.front

        print("Queue:", end=" ")

        while curr is not None:
            print(curr.data, end=" ")

            curr = curr.next

        print()


# Create Queue
q = Queue()

# Enqueue elements
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)

# Display
q.display()

# Peek
print("Front element:", q.peek())

# Dequeue
print("Removed:", q.dequeue())
print("Removed:", q.dequeue())

# Display after deletion
q.display()

# Peek again
print("Front element:", q.peek())

# Check empty
print("Is queue empty?", q.is_empty())
```
