## Doubly Linked List in Detail

A doubly linked list is a type of linked list where each node has three parts:

1.  **Data**: The actual value or information stored in the node.
2.  **`next` pointer**: A pointer (or reference) to the next node in the sequence.
3.  **`prev` pointer**: A pointer (or reference) to the previous node in the sequence.

Unlike a singly linked list, which only allows traversal in one direction (forward), a doubly linked list allows traversal in both directions (forward and backward) because each node maintains pointers to both its successor and predecessor.

### Structure of a Doubly Linked List Node

Each node typically looks like this:

```
+------+-------+------+
| prev | data  | next |
+------+-------+------+
```

-   `prev`: Points to the node immediately preceding it.
-   `data`: Stores the actual element.
-   `next`: Points to the node immediately succeeding it.

### Key Components of a Doubly Linked List

1.  **Head**: A pointer to the first node in the list. If the list is empty, the head pointer is usually `None` or `null`.
2.  **Tail**: A pointer to the last node in the list. This makes operations like appending to the end more efficient than in a singly linked list.

### Advantages of Doubly Linked Lists

1.  **Bidirectional Traversal**: You can traverse the list in both forward and backward directions, which is not possible with singly linked lists without external mechanisms.
2.  **Efficient Deletion**: Deleting a node is more efficient because you have direct access to the previous node, making it easy to update pointers. In a singly linked list, to delete a node, you often need to traverse from the head to find its predecessor.
3.  **Efficient Insertion**: Inserting a new node before a given node is also more straightforward and efficient.
4.  **Tail Access**: If a `tail` pointer is maintained, appending to the end of the list takes O(1) time.

### Disadvantages of Doubly Linked Lists

1.  **Increased Memory Usage**: Each node requires an additional pointer (`prev`), meaning it consumes more memory than a singly linked list.
2.  **More Complex Implementation**: Operations like insertion and deletion involve managing two pointers (`next` and `prev`) for each node, making the implementation slightly more complex and prone to errors compared to singly linked lists.

### Common Operations and Their Time Complexity

Assuming the head (and often tail) is known:

-   **Insertion at the Beginning**: O(1)
-   **Insertion at the End**: O(1) (if tail pointer is maintained)
-   **Insertion After a Given Node**: O(1)
-   **Deletion at the Beginning**: O(1)
-   **Deletion at the End**: O(1) (if tail pointer is maintained)
-   **Deletion of a Given Node**: O(1) (if a reference to the node is given)
-   **Search for an Element**: O(n) (where n is the number of nodes)

# Insert At Begining

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


# Create nodes
a = Node(10)
b = Node(20)
c = Node(30)


# Connect nodes
a.next = b
b.prev = a

b.next = c
c.prev = b


# Head
head = a


# Traverse forward BEFORE insertion
print("Before insertion:")

curr = head

while curr is not None:
    print(curr.data, end=" ")
    curr = curr.next

print()


# Insert at beginning
def insert_at_beg(head, data):

    # Create new node
    new_node = Node(data)

    # If list is empty
    if head is None:
        return new_node

    # New node points to old head
    new_node.next = head

    # Old head points back to new node
    head.prev = new_node

    # New node becomes the new head
    return new_node


# Insert 5 at beginning
head = insert_at_beg(head, 5)


# Traverse forward AFTER insertion
print("After insertion:")

curr = head

while curr is not None:
    print(curr.data, end=" ")
    curr = curr.next

print()
````

# Insert at kth index

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


# Insert at kth index
def insert_at_kth(head, data, k):

    # Create new node
    new_node = Node(data)

    # Case 1: Empty list
    if head is None:
        return new_node

    # Case 2: Insert at beginning
    if k == 0:
        new_node.next = head
        head.prev = new_node
        return new_node

    # Traverse to the kth node
    curr = head

    for i in range(k):
        curr = curr.next

        # Position is outside the list
        if curr is None:
            return head

    # curr is now the node currently at index k

    # Connect new node with previous node
    new_node.prev = curr.prev
    curr.prev.next = new_node

    # Connect new node with current node
    new_node.next = curr
    curr.prev = new_node

    return head


# Create nodes
a = Node(10)
b = Node(20)
c = Node(30)
d = Node(40)

# Connect nodes
a.next = b
b.prev = a

b.next = c
c.prev = b

c.next = d
d.prev = c

# Head
head = a


# Before insertion
print("Before:")
curr = head

while curr is not None:
    print(curr.data, end=" ")
    curr = curr.next

print()


# Insert 25 at index 2
head = insert_at_kth(head, 25, 2)


# After insertion
print("After:")
curr = head

while curr is not None:
    print(curr.data, end=" ")
    curr = curr.next

print()
````

# Insert at End

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


def insert_at_end(head, data):

    # Create new node
    new_node = Node(data)

    # If list is empty
    if head is None:
        return new_node

    # Traverse to the last node
    curr = head

    while curr.next is not None:
        curr = curr.next

    # Connect last node to new node
    curr.next = new_node

    # Connect new node back to last node
    new_node.prev = curr

    # Return head
    return head


# Create nodes
a = Node(10)
b = Node(20)
c = Node(30)

# Connect nodes
a.next = b
b.prev = a

b.next = c
c.prev = b

# Head
head = a


# Before insertion
print("Before:")
curr = head

while curr is not None:
    print(curr.data, end=" ")
    curr = curr.next

print()


# Insert 40 at end
head = insert_at_end(head, 40)


# After insertion
print("After:")
curr = head

while curr is not None:
    print(curr.data, end=" ")
    curr = curr.next

print()
````
