# 1. Delete from Beginning

```python
class Node:
  def __init__(self, data):
        self.data = data
        self.next = None
a=Node(75)
b=Node(85)
c=Node(95)

a.next=b
b.next=c

head=a
# Delete from beginning
head = head.next

def print_list(head):
  curr=head

  while curr is not None:
    print(curr.data,end=" ")
    curr=curr.next

  print()

print_list(head)
```

Let's break down each code snippet:

### Understanding the `Node` class and `print_list` function

Before diving into deletions, it's important to understand the basic building blocks:

*   **`class Node:`**: This class defines a node in a linked list. Each node has two parts:
    *   `self.data`: Stores the actual value or data of the node (e.g., 75, 85, 95).
    *   `self.next`: A pointer (or reference) to the next node in the sequence. It's initialized to `None` if it's the last node.

*   **`def print_list(head):`**: This is a helper function used to traverse the linked list and print the `data` of each node. It starts from the `head` (the first node) and moves through `curr = curr.next` until it reaches a node whose `next` is `None` (the end of the list).

### 1. Delete from Beginning (from cell `aff8JaTKitVa`)

```python
class Node:
  def __init__(self, data):
        self.data = data
        self.next = None
a=Node(75)
b=Node(85)
c=Node(95)

a.next=b
b.next=c

head=a
# Delete from beginning
head = head.next

def print_list(head):
  curr=head

  while curr is not None:
    print(curr.data,end=" ")
    curr=curr.next

  print()

print_list(head)
```

#### Explanation:

1.  **List Initialization**: Three nodes `a`, `b`, and `c` are created with data 75, 85, and 95 respectively. They are linked together: `a -> b -> c`. The `head` initially points to `a` (75).
2.  **`head = head.next`**: This is the core line for deleting the first element. By moving the `head` pointer to the *next* node in the list (`b` in this case), the original first node (`a`) is effectively unlinked from the list. Python's garbage collector will automatically reclaim the memory for node `a` if no other references to it exist.
3.  **`print_list(head)`**: The `print_list` function is then called, which prints the remaining elements starting from the new `head` (node `b`).

**Output**: `85 95` (75 has been deleted).

### 2. Delete from Last (from cell `ndpcNhPelX3v`)

```python
class Node:
  def __init__(self, data):
        self.data = data
        self.next = None

a = Node(75)
b = Node(85)
c = Node(95)

a.next = b
b.next = c

head = a

#Delete last Node
curr=head

while curr.next.next is not None:
  curr=curr.next
curr.next=None

def print_list(head):
  curr=head
  while curr is not None:
    print(curr.data,end=" ")
    curr=curr.next
  print()


print_list(head)
```

#### Explanation:

1.  **List Initialization**: Similar to the first example, a list `a -> b -> c` is created, with `head` pointing to `a` (75).
2.  **`curr = head`**: A temporary pointer `curr` is set to the `head`.
3.  **`while curr.next.next is not None:`**: This loop iterates until `curr` points to the *second-to-last* node. Let's trace:
    *   Initially, `curr` is `a`. `a.next` is `b`, `b.next` is `c`. `c` is not `None`, so `curr.next.next` (which is `c.next`) is `None`. Thus, `curr.next.next is not None` is `True` (if `c.next` was `None`, the loop would stop one node earlier. The condition `curr.next.next is not None` correctly identifies the node *before* the last one. For `a->b->c`, when `curr` is `a`, `curr.next.next` is `c.next`, which is `None`. So `curr.next.next is not None` is `False`. This loop condition is designed to stop `curr` at the node *before* the one you want to delete. In this case, to delete `c`, `curr` should stop at `b`.
    *   Revised trace: `curr` starts at `a`. The condition `curr.next.next is not None` evaluates to `b.next is not None` (which is `c is not None`) if `curr` is `a`. This means `curr.next.next` is `None` (for `c.next`). Let's re-evaluate the loop: `a.next.next` means `b.next` which is `c`. `c.next` is `None`. So `curr.next.next` (which is `c.next`) is `None`. The condition `curr.next.next is not None` will be `False` when `curr` is `a` for the list `a -> b -> c`. This means `curr` will stay at `a`, which is incorrect for deleting the last node.

    *   **Correction needed**: The `while` loop condition `curr.next.next is not None` is correct for stopping `curr` at the node *before* the last node, but only if the list has at least three nodes. If the list is `a -> b -> c`:
        *   `curr = a`
        *   `a.next.next` (i.e., `b.next`) is `c`. `c.next` is `None`. So `curr.next.next` is `None`. The condition `curr.next.next is not None` is `False`. The loop does not execute.
        *   This means `curr` remains `a`. `curr.next = None` would delete `b`, not `c`. This code has a bug for a list of 3 nodes.

    *   **Correct logic for deleting last node**: The `curr` pointer should stop at the node *before* the last node. This means its `next` pointer should point to the last node. So the condition should be `while curr.next.next is not None:` to stop `curr` when `curr.next` is the second to last node.
    *   Let's consider `a -> b -> c`. We want `curr` to be `b`.
        *   `curr = a`. Is `a.next.next` (`b.next`) not `None`? Yes, `c` is not `None`. So `curr` becomes `b`.
        *   `curr = b`. Is `b.next.next` (`c.next`) not `None`? No, `c.next` is `None`. So the loop stops.
        *   `curr` is now `b`. This is correct.

4.  **`curr.next = None`**: Once `curr` is the second-to-last node, setting `curr.next = None` breaks the link to the last node. The last node is now unreferenced and effectively deleted.
5.  **`print_list(head)`**: Prints the modified list.

**Output**: `75 85` (95 has been deleted).

### 3. Delete from Kth Index (from cell `hTs5dxN-oHQB`)

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


a = Node(75)
b = Node(85)
c = Node(95)
d = Node(105)

a.next = b
b.next = c
c.next = d

head = a

# Delete index 2
index = 2

curr = head

for i in range(index - 1):
    curr = curr.next

curr.next = curr.next.next


def print_list(head):
    curr = head

    while curr is not None:
        print(curr.data, end=" ")
        curr = curr.next

    print()


print_list(head)
```

#### Explanation:

1.  **List Initialization**: Four nodes `a`, `b`, `c`, and `d` are created with data 75, 85, 95, and 105 respectively. They are linked: `a -> b -> c -> d`. The `head` points to `a` (75).
2.  **`index = 2`**: Specifies that the node at index 2 (0-based indexing, so this is the third node, which contains 95) should be deleted.
3.  **`for i in range(index - 1): curr = curr.next`**: This loop moves the `curr` pointer to the node *before* the one to be deleted. If `index` is 2, the loop runs for `i = 0` and `i = 1`:
    *   `curr` starts at `a`.
    *   `i = 0`: `curr` moves to `b`.
    *   `i = 1`: `curr` moves to `c`. No, this should be incorrect. `curr` should point to the node at `index - 1`. If `index` is 2, `index - 1` is 1. So `curr` should stop at index 1, which is node `b`.

    *   Let's retrace: `index = 2`. `range(index - 1)` means `range(1)`, so `i` will be `0`.
        *   `curr` starts at `head` (`a`).
        *   Loop (for `i=0`): `curr = curr.next` (`curr` becomes `b`).
        *   Loop ends.
        *   So, `curr` points to node `b` (the node *before* the node at index 2).

4.  **`curr.next = curr.next.next`**: This is the deletion step. `curr.next` currently points to the node to be deleted (node `c` with data 95). `curr.next.next` points to the node *after* the one to be deleted (node `d` with data 105). By setting `curr.next` to `curr.next.next`, node `b` now directly points to node `d`, effectively bypassing and unlinking node `c`.
5.  **`print_list(head)`**: Prints the resulting list.

**Output**: `75 85 105` (95 has been deleted).
