A circular linked list is a variation of a singly linked list where the last node points back to the first node, forming a circle. Unlike a traditional linked list, there is no `null` terminator at the end. This structure allows for continuous traversal of the list without encountering an end.

Here are some key characteristics and details:

*   **Circular Nature**: The most defining characteristic is that the `next` pointer of the last node points to the first node of the list. This creates a loop.
*   **No Null**: There is no `null` pointer to mark the end of the list, which means that any node can be a starting point for traversal.
*   **Head Node**: While any node can be a starting point, a `head` (or `tail`) pointer is usually maintained to easily access the beginning (or end) of the list.
*   **Traversal**: To traverse a circular linked list, you start at a given node and continue following the `next` pointers until you return to the starting node. You need a condition to stop the traversal, typically checking if the current node is equal to the starting node.

**Advantages of Circular Linked Lists:**

1.  **Continuous Traversal**: Useful for applications where you need to repeatedly cycle through items, such as in a round-robin scheduling algorithm.
2.  **No Null Checks**: Simplifies some operations, as you don't need to check for `null` at the end of the list during traversal.
3.  **Efficient Access to Both Ends (if using a tail pointer)**: If you maintain a `tail` pointer, both the first and last elements can be accessed in O(1) time. The `tail.next` will point to the head.

**Disadvantages of Circular Linked Lists:**

1.  **Infinite Loop Potential**: Traversal logic needs careful handling to avoid infinite loops, as there's no natural end point.
2.  **More Complex Operations**: Operations like finding the middle element or reversing the list can be more complex than in a singly linked list.

**Common Operations:**

*   **Insertion**:
    *   **At the beginning**: Create a new node, make its `next` point to the current head, and make the last node's `next` point to the new node. Update the head.
    *   **At the end**: Create a new node, make the current last node's `next` point to the new node, and the new node's `next` point to the head. Update the last node (or tail).
*   **Deletion**:
    *   Requires finding the node to be deleted and its predecessor. The predecessor's `next` pointer is then updated to skip the deleted node. Special care is needed if deleting the head or the only node.
*   **Traversal**: Start from the head, print the data, and move to the next node until you reach the head again.

**Use Cases:**

*   **Operating Systems**: Used in CPU scheduling (e.g., Round Robin scheduling), where processes are stored in a circular queue.
*   **Music Players**: Implementing playlists where songs play in a loop.
*   **Game Development**: Managing game entities that need to be processed in a cycle.
*   **Buffer Management**: Creating circular buffers for data streams.

```
+-----+     +-----+     +-----+     +----+
  ^                                     |
  |-------------------------------------+
```

# 1. Circular Linked List — Basic Creation & Traversal

````python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

#create Nodes
a=Node(10)
b=Node(20)
c=Node(30)
d=Node(40)

a.next=b
b.next=c
c.next=d
d.next=a

head=a

curr=head

while True:
  print(curr.data,end=" ")
  curr=curr.next

  if curr==head:
    break
````

# 2. Insert at Beginning

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


a = Node(10)
b = Node(20)
c = Node(30)

a.next = b
b.next = c
c.next = a

head = a

new_node=Node(90)

#Find last Node

curr=head
while curr.next!=head:
  curr=curr.next

#Insert at beg
new_node.next=head
curr.next=new_node
head=new_node

# Display
curr = head

while True:
    print(curr.data, end=" ")
    curr = curr.next

    if curr == head:
        break
````

# 3. Insert at End

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


a = Node(10)
b = Node(20)
c = Node(30)

a.next = b
b.next = c
c.next = a

head = a


# New node
new_node = Node(40)

# Find last node
curr = head

while curr.next != head:
    curr = curr.next

# Insert at end
curr.next = new_node
new_node.next = head


# Display
curr = head

while True:
    print(curr.data, end=" ")
    curr = curr.next

    if curr == head:
        break
````
