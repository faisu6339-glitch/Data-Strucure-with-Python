#Bubble Sort Algorithm

```python
def bubble_sort(arr):
  n=len(arr)

  for i in range(n):
    swapped=False
    for j in range(0,n-i-1):
      if arr[j] > arr[j+1]:
        arr[j],arr[j+1]=arr[j+1],arr[j]
        swapped=True
    if not swapped:
      break
  return arr

my_arr=[12,85,92,74,365,8]
result=bubble_sort(my_arr)
print(result)


```

```python

def bubble_sort(names):
    n = len(names)

    for i in range(n):
        for j in range(n - i - 1):
            if names[j] > names[j + 1]:
                names[j], names[j + 1] = names[j + 1], names[j]

    return names


employees = ["Zainab Fatma", "Aaliya Fatma","Faisal Sheikh" ,"Talat", "Noor Aisha", "Ruhul Mateen Ahmad"]

print("Original List:", employees)

sorted_names = bubble_sort(employees)

print("Sorted List:", sorted_names)
```

#Selection Sort Algorithm

```python
def selection_sort(arr):
  n=len(arr)
  for i in range(n-1):
    min_idx=i
    for j in range(i+1,n):
       if arr[j]<arr[min_idx]:
          min_idx=j
    if min_idx!=i:
        arr[i],arr[min_idx]=arr[min_idx],arr[i]
  return arr

  my_list = [64, 25, 12, 22, 11]
print(f"Original list: {my_list}")
sorted_list = selection_sort(my_list)
print(f"Sorted list: {sorted_list}")

my_list_2 = [5, 4, 3, 2, 1]
print(f"Original list 2: {my_list_2}")
sorted_list_2 = selection_sort(my_list_2)
print(f"Sorted list 2: {sorted_list_2}")

```

```python
products = [
    ("Laptop", 65000),
    ("Mouse", 700),
    ("Keyboard", 1500),
    ("Monitor", 12000),
    ("Printer", 8500)
]

def selection_sort_price(data):
  n=len(data)
  for i in range(n):
    min_ind=i
    for j in range(i+1,n):
      if data[j][1]<data[min_ind][1]:
        min_ind=j
    if min_ind!=i:
      data[i],data[min_ind]=data[min_ind],data[i]
  return data


sorted_products = selection_sort_price(products)

for name, price in sorted_products:
    print(name, price)
```

```python
files = [
    ("movie.mp4", 800),
    ("photo.png", 5),
    ("music.mp3", 12),
    ("notes.pdf", 2),
    ("project.zip", 150)
]


def selection_sort_size(data):
    n = len(data)

    for i in range(n):
        min_index = i

        for j in range(i + 1, n):
            if data[j][1] < data[min_index][1]:
                min_index = j

        data[i], data[min_index] = data[min_index], data[i]

    return data


print(selection_sort_size(files))
```
