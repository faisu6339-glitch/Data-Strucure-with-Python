#**Binary Search**

##### **Explanation**

Binary Search is the absolute purest manifestation of the Divide-and-Conquer paradigm applied to search operations. When presented with a strictly sorted array, scanning linearly from start to finish is a massive waste of computational resources. Binary Search leverages the sorted nature of the data to eliminate exactly half of the remaining search space with every single comparison operation. It achieves this by inspecting the exact middle element of the current boundaries.

If the middle element is mathematically smaller than the target, the algorithm knows with absolute certainty that the target cannot exist in the left half of the array. It forcefully moves the left boundary past the midpoint. Conversely, if the middle element is larger, it discards the entire right half. This aggressive pruning reduces a linear time search operation down to a highly optimal logarithmic time operation, allowing a search through one billion elements to complete in roughly thirty steps.

The purpose of Binary Search is to provide a foundation for searching massive, static datasets rapidly. It introduces the critical concept of maintaining "pointers" or "boundaries" that dynamically shift as the problem space is recursively or iteratively conquered. Understanding this boundary manipulation is a mandatory prerequisite for all advanced divide-and-conquer algorithms.

##### **Understanding Concept**
Imagine opening a heavy dictionary to find a specific word. You do not start reading from page one. You open the book roughly to the exact center. If the word you are looking for comes alphabetically after the page you are currently on, you completely ignore the entire first half of the dictionary. You then split the remaining second half in the middle again. You repeat this halving process until you land on the exact page you need. You divided the book, conquered the halves, and found your target instantly.

##### **How it Works**
- Step One: Verify that the input array is strictly sorted in ascending or descending order. If it is unsorted, Binary Search will fail catastrophically.
- Step Two: Establish a left boundary pointer at the absolute beginning of the array and a right boundary pointer at the absolute end.
- Step Three: Enter a loop that continues strictly as long as the left boundary is less than or equal to the right boundary.
- Step Four: Calculate the middle index by adding the left and right boundaries and mathematically performing integer division by two.
- Step Five: Compare the element at the middle index against the target value.
- Step Six: If the middle element perfectly matches the target, return the middle index to halt the execution.
- Step Seven: If the target is strictly greater than the middle element, update the left boundary pointer to the middle index plus one.
- Step Eight: If the target is strictly less than the middle element, update the right boundary pointer to the middle index minus one.
- Step Nine: If the loop exhausts the entire search space without a match, return an indicator signifying the target does not exist.

```python
# Problem:
# Given a sorted array, return the index if the target exists.
# If the target does not exist, return the index where it should be inserted
# to keep the array sorted.

def search_insert(arr, target):
    left, right = 0, len(arr) - 1

    while left <= right:
        mid = left + (right - left) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return left


arr = [1, 3, 5,6]
target = 4

print(search_insert(arr, target))
```

```python
# moving zeroes to the end
def move_zeros(arr):
    left = 0

    for right in range(len(arr)):
        if arr[right] != 0:
            arr[left], arr[right] = arr[right], arr[left]
            left += 1

    return arr


arr = [0, 1, 0, 3, 12]
print(move_zeros(arr))
```

```python
def move_zeros(arr):
    count = 0

    for i in range(len(arr)):
        if arr[i] != 0:
            arr[count] = arr[i]
            count += 1

    while count < len(arr):
        arr[count] = 0
        count += 1

    return arr


arr = [0, 1, 0, 3, 12]
print(move_zeros(arr))
```

```python
#Find minimum
def find_min(arr):
    left, right = 0, len(arr) - 1

    while left < right:
        mid = left + (right - left) // 2

        if arr[mid] > arr[right]:
            # Minimum is in the right half
            left = mid + 1
        else:
            # Minimum is at mid or in the left half
            right = mid

    return arr[left]


arr = [4, 5, 6, 7, 0, 1, 2]
print(find_min(arr))
```
