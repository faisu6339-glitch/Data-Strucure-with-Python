#**Linked List**

##### **Explanation**
A Singly Linked List is a fundamental linear data structure consisting of a sequence of elements called nodes. Unlike traditional arrays, which demand a single, contiguous block of memory, the nodes of a Linked List are scattered throughout the computer's memory space. To maintain sequential order, each node acts as an independent container storing two critical pieces of information: the actual data value, and a directional reference (or pointer) to the exact memory address of the subsequent node in the sequence.

The entire list is anchored by a single entry point known as the Head pointer. The Head does not store data itself; it strictly stores the memory address of the very first node. To read or access the third item in the list, the computer must start at the Head, travel to the first node, follow its pointer to the second node, and finally follow the second node's pointer to the third. The sequence terminates when a node's pointer references an empty or null value, designating it as the Tail.

This architectural design exists to solve the rigid resizing limitations of contiguous arrays. Inserting a new value at the beginning of an array forces the computer to shift every subsequent element one space to the right, triggering heavy processing delays. In a Singly Linked List, inserting a new element at the beginning requires zero shifting; it simply requires creating a new node and updating two memory pointers. This grants Linked Lists unparalleled O(1) efficiency for insertions and deletions at known memory locations.

##### **Understanding Concept**
Imagine organizing a scavenger hunt for a group of friends. You do not place all the clues in one single, organized row on a table (which represents an array). Instead, you hide the first clue in the mailbox. When your friends read the first clue, it contains a message (the data) and specifically tells them to look under the doormat for the next clue (the pointer). The clue under the doormat leads to the garage, and the garage leads to the attic. The scavenger hunt represents a Linked List. The players can only find the third clue by physically traveling to the first and second clues sequentially. If someone accidentally destroys the clue in the garage, the location of the attic clue is lost forever.

##### **How it Works**
Step One: Define a Node class that initializes with a data value and a next pointer set to null.

Step Two: Define the LinkedList class, initializing a Head attribute that points to null, representing an empty list.

Step Three: To insert a new node at the front, create the node, point its next reference to the current Head, and update the Head to point to the new node.

Step Four: To insert a node at the end, verify if the list is empty. If empty, assign the node to the Head. Otherwise, traverse the entire list starting from the Head until reaching the node whose next pointer is null, then reassign that pointer to the new node.

Step Five: To traverse the list, create a temporary current pointer starting at the Head. Extract the data, then update the current pointer to equal current.next. Loop this process until current becomes null.

Edge Cases: A robust implementation must gracefully handle operations on completely empty lists (Head is null), operations targeting the very first node, and operations targeting the absolute final node.

**Real Life Example**

Imagine a scavenger hunt where each clue leads to the location of the next clue.
The first clue may be in the mailbox, which tells players to check under the doormat next.
The clue under the doormat then points to the garage, and the garage points to the attic.
Players must follow the clues one by one because they cannot directly jump to the last clue.
If one clue is lost or destroyed, the path to the remaining clues also gets broken.
A Linked List works in the same way, where each node stores data and the address of the next node.

````python
# Define the Node class to represent an individual element in memory
class Node:

    # Initialize the node object
    def __init__(self, data):

        # Store data inside the node
        self.data = data

        # Initialize next pointer
        self.next = None


# Define the Singly Linked List class
class SinglyLinkedList:

    # Initialize the linked list
    def __init__(self):

        # Initially the list is empty
        self.head = None

    # Append element at the end
    def append(self, data):

        # Create new node
        new_node = Node(data)

        # Check if list is empty
        if self.head is None:
            self.head = new_node
            return

        # Traverse to last node
        current = self.head

        while current.next is not None:
            current = current.next

        # Attach new node at end
        current.next = new_node

    # Insert element at beginning
    def prepend(self, data):

        # Create new node
        new_node = Node(data)

        # Point new node to current head
        new_node.next = self.head

        # Move head to new node
        self.head = new_node

    # Display linked list
    def display(self):

        elements = []

        current = self.head

        while current is not None:
            elements.append(current.data)
            current = current.next

        return elements


# ==================================================
# Example 1
# Input: append(10), append(20)
# Output: [10, 20]
# Explanation: The list is empty initially. Node(10) is attached directly to the Head. Next, the algorithm traverses to Node(10) and rigorously assigns Node(10).next to equal Node(20).
# ==================================================

print("Example 1")

list1 = SinglyLinkedList()

list1.append(10)
list1.append(20)

print("Output:", list1.display())


# ==================================================
# Example 2
# Input: append(5), prepend(1)
# Output: [1, 5]
# Explanation: Node(5) becomes the Head. The prepend function generates Node(1), assigns Node(1).next to explicitly target Node(5), and then securely overwrites the Head pointer to target Node(1).
# ==================================================

print("\nExample 2")

list2 = SinglyLinkedList()

list2.append(5)
list2.prepend(1)

print("Output:", list2.display())


# ==================================================
# Example 3
# Input: Empty List
# Output: []
# Explanation: The Head pointer remains explicitly assigned to None. The display loop immediately encounters the None object, violently bypassing the iteration block and securely returning the empty tracking array.
# ==================================================

print("\nExample 3")

list3 = SinglyLinkedList()

print("Output:", list3.display())
````



**Time Complexity**
Best case: O(1) for prepending an element directly at the Head, as no traversal is required.
Average case: O(N) for accessing, searching, or appending an element to the Tail (without a Tail pointer).
Worst case: O(N) for traversing the entire list to locate an element residing at the very end.
Explanation: Singly Linked Lists mathematically lack random access. Finding an element at index K strictly requires executing K sequential pointer hops. However, inserting an element at a known pointer location securely resolves in constant O(1) time because only two pointer assignments are modified.

**Space Complexity**
Explanation: O(N) scaling spatial complexity. The data structure rigorously allocates a new object in the system heap memory for every single element added. Each object contains the data value and an overhead memory allocation specifically holding the reference pointer.

# **Doubly Linked List**

##### **Explanation**
A Doubly Linked List is an advanced architectural upgrade to the standard Singly Linked List paradigm. While a singly linked node only contains a memory pointer directing execution forward to the subsequent node, a doubly linked node inherently contains three isolated components: the stored data, a pointer targeting the next consecutive node, and a crucial new pointer explicitly targeting the previous preceding node in the sequence.

This dual-pointer architecture fundamentally destroys the unidirectional limitation of basic linked lists. Programs can safely and efficiently traverse the entire data structure both forward from the Head and backward from the Tail. If an algorithm maintains a direct pointer to a specific node located deep inside the list, it can delete that node perfectly in O(1) time. In a Singly Linked List, deleting a known node still heavily requires traversing from the Head simply to locate and modify the preceding node's pointer. The previous pointer completely eliminates this traversing penalty.

The mandatory trade-off for this massive leap in bidirectional navigational efficiency is increased spatial complexity. Every single node now consumes additional system memory to actively store the secondary previous pointer. Furthermore, the implementation logic escalates in complexity, as every single insertion or deletion operation now strictly requires securely updating four disparate memory pointers instead of two.

##### **Understanding Concept**
Imagine a massive, interconnected passenger train. In a Singly Linked List train, every single train car only has a door on the front end pointing toward the engine. If you are sitting in car number five and drop your ticket in car number four, you cannot simply walk backward. You must exit the train completely, walk all the way to the main engine at the front, and walk forward sequentially through cars one, two, and three just to reach four.

A Doubly Linked List train adds doors to both the front and the back of every single car. If you are in car five, you simply open the back door and step directly backward into car four. The structural connections explicitly support travel in both absolute directions seamlessly.

##### **How it Works**
Step One: Define an upgraded Node class requiring a data value, a next pointer initialized to null, and a prev pointer initialized to null.

Step Two: Define the DoublyLinkedList class maintaining a Head pointer. Maintaining a dedicated Tail pointer is also highly recommended to permit O(1) tail insertions and immediate reverse traversals.

Step Three: To append an element, verify if the list is empty. If empty, the new node becomes both Head and Tail.

Step Four: If not empty, take the current Tail node. Point the current Tail's next reference to the new node. Crucially, point the new node's prev reference backward to the current Tail.

Step Five: Update the global Tail pointer to officially target the newly inserted node.

Edge Cases: Removing the sole remaining node in the list rigorously requires updating both the Head and Tail pointers to explicitly return to the null state.

**Real Life Example**

Imagine a long passenger train where each train car is connected to another.
In a Singly Linked List, every car has a door only toward the front, so passengers can move only forward.
If someone in car five wants to go back to car four, they must start again from the engine and move forward step by step.
In a Doubly Linked List, each car has doors at both the front and back ends.
This allows passengers to move easily forward or backward between cars without restarting the journey.
Similarly, singly linked lists support one-way traversal, while doubly linked lists allow movement in both directions.

````python
# Define the DoublyNode class
class DoublyNode:

    # Initialize node
    def __init__(self, data):

        # Store data
        self.data = data

        # Forward pointer
        self.next = None

        # Backward pointer
        self.prev = None


# Define the Doubly Linked List class
class DoublyLinkedList:

    # Initialize linked list
    def __init__(self):

        # Head pointer
        self.head = None

        # Tail pointer
        self.tail = None

    # Append node at end
    def append(self, data):

        # Create new node
        new_node = DoublyNode(data)

        # If list is empty
        if self.head is None:

            self.head = new_node
            self.tail = new_node

            return

        # Link current tail to new node
        self.tail.next = new_node

        # Link new node back to tail
        new_node.prev = self.tail

        # Move tail pointer
        self.tail = new_node

    # Display list in reverse order
    def display_backward(self):

        elements = []

        # Start from tail
        current = self.tail

        # Traverse backward
        while current is not None:

            elements.append(current.data)

            current = current.prev

        return elements


# ==================================================
# Example 1
# Input: append(1), append(2)
# Output: [2, 1]
# Explanation: The Head locks to 1, the Tail locks to 2. Node(2).prev targets Node(1). The display function starts at the Tail (2), records it, follows the prev pointer to 1, records it, and terminates safely upon encountering None.
# ==================================================

print("Example 1")

list1 = DoublyLinkedList()

list1.append(1)
list1.append(2)

print("Output:", list1.display_backward())


# ==================================================
# Example 2
# Input: append(10)
# Output: [10]
# Explanation: Both Head and Tail pointers target Node(10). The reverse loop executes exactly one iteration, extracts the value, follows the prev pointer resolving to None, and exits the loop completely.
# ==================================================

print("\nExample 2")

list2 = DoublyLinkedList()

list2.append(10)

print("Output:", list2.display_backward())


# ==================================================
# Example 3
# Input: Empty List
# Output: []
# Explanation: The Tail pointer equals None. The traversal pointer receives this null state. The while loop condition immediately fails, bypassing iteration and returning an empty list.
# ==================================================

print("\nExample 3")

list3 = DoublyLinkedList()

print("Output:", list3.display_backward())
````

**Time Complexity**
Best case: O(1) for appending explicitly to the Tail, bypassing traversal constraints entirely.
Average case: O(N) for locating a target node buried deeply in the center of the list.
Worst case: O(N) evaluating the entire length of the structure sequentially.
Explanation: While bidirectional traversal does not fundamentally alter the Big-O searching complexity, it practically halves the execution time if you heavily optimize the search protocol to initiate from the Tail when evaluating target indices resting in the back half of the list.

**Space Complexity**
Explanation: O(N) allocation boundaries. While it scales linearly precisely like a Singly Linked List, the hidden constant overhead is significantly heavier. The operating system must forcefully allocate an extra pointer object inside every single node generated, consuming measurable RAM on large-scale databases.




## **Circular Linked List**

### **Explanation**
A Circular Linked List is a variation of a Linked List where the final node points back to the head node instead of `None`. This creates a continuous loop structure allowing repeated traversal without termination.

Circular Linked Lists are useful when cyclic processing is required. Unlike traditional Singly Linked Lists, traversal naturally loops back to the beginning automatically after reaching the final node.

### **Understanding Concept**
Imagine people standing in a circle while holding hands. The final person connects back to the first person instead of stopping the chain. If you continue walking around the circle, you repeatedly return to the same starting point.

A Circular Linked List works exactly the same way because the final node reconnects to the head node.

### **How it Works**
Step 1: Create nodes containing data and a next pointer.

Step 2: Maintain a head pointer storing the first node.

Step 3: Ensure the final node points back to the head node instead of `None`.

Step 4: During traversal, stop when the traversal pointer reaches the head node again.

Step 5: Carefully manage insertion and deletion because there is no null tail marker.



### **Real-World Example**
A music playlist operating in repeat mode behaves exactly like a Circular Linked List. After the final song finishes, playback automatically returns to the first song instead of stopping.

Operating systems also use Circular Linked Lists in round-robin CPU scheduling systems.

````python
# Node class
class Node:

    # Constructor
    def __init__(self, data):

        # Store value
        self.data = data

        # Initialize next pointer
        self.next = None


# Circular Linked List class
class CircularLinkedList:

    # Constructor
    def __init__(self):

        # Initialize head
        self.head = None

    # Insert node at end
    def append(self, data):

        # Create new node
        new_node = Node(data)

        # Empty list case
        if not self.head:

            self.head = new_node

            # Point node to itself
            new_node.next = self.head

            return

        # Traverse to last node
        current = self.head

        while current.next != self.head:
            current = current.next

        # Insert new node
        current.next = new_node

        # Connect back to head
        new_node.next = self.head

    # Display list
    def display(self):

        # Empty list check
        if not self.head:
            print("[]")
            return

        current = self.head

        while True:

            print(current.data, end=" -> ")

            current = current.next

            # Stop after complete cycle
            if current == self.head:
                break

        print("back to head")


# ==================================================
# Example 1
# Input: 10 -> 20 -> 30
# Output: 10 -> 20 -> 30 -> back to head
# Explanation: The final node reconnects directly to the head node, forming a continuous circular chain.
# ==================================================

print("Example 1")

circular_list1 = CircularLinkedList()

circular_list1.append(10)
circular_list1.append(20)
circular_list1.append(30)

circular_list1.display()


# ==================================================
# Example 2
# Input: Single node containing 5
# Output: 5 -> back to head
# Explanation: The single node points to itself, creating a valid circular linked list structure.
# ==================================================

print("\nExample 2")

circular_list2 = CircularLinkedList()

circular_list2.append(5)

circular_list2.display()
````



**Time Complexity**
Insertion at End: O(N)

Traversal: O(N)

Search: O(N)

**Space Complexity**
O(N)

### **Activity 1: Detect a Cycle (Tortoise and Hare)**

**Problem:**  
An insidious bug caused the Tail node of a Singly Linked List to point its `next` reference back to the Head node, creating an infinite loop. Modify the list to include a method that detects this cycle without modifying node data or using extra hash maps.


````python
# Node class
class Node:

    # Constructor
    def __init__(self, data):
        self.data = data
        self.next = None


# Linked List class
class LinkedList:

    # Constructor
    def __init__(self):
        self.head = None

    # Detect cycle using Floyd's Algorithm
    def has_cycle(self):

        # Initialize slow and fast pointers
        slow = self.head
        fast = self.head

        # Traverse list
        while fast and fast.next:

            # Move slow pointer by one step
            slow = slow.next

            # Move fast pointer by two steps
            fast = fast.next.next

            # Cycle detected
            if slow == fast:
                return True

        return False


# Create linked list
linked_list = LinkedList()

# Create nodes
first = Node(1)
second = Node(2)
third = Node(3)

# Connect nodes
linked_list.head = first
first.next = second
second.next = third

# Create cycle
third.next = first

# Detect cycle
print(linked_list.has_cycle())
````

**Answer:**  
Use Floyd’s Cycle Detection Algorithm with a slow pointer and a fast pointer.

**Explanation:**  
The slow pointer moves one node at a time while the fast pointer moves two nodes at a time.

If a cycle exists, the fast pointer eventually catches the slow pointer inside the loop.

This approach detects cycles efficiently without modifying node values or using additional memory-heavy data structures.

**Time Complexity:**  
O(n)

**Space Complexity:**  
O(1)


### **Activity 2: Implement Singly Node Deletion by Value**

**Problem:**  
Expand the standard Singly Linked List class to include a `delete_by_value(target)` method. The method must locate a node containing the target value and remove it from the list safely.


````python
# Node class
class Node:

    # Constructor
    def __init__(self, data):
        self.data = data
        self.next = None


# Linked List class
class LinkedList:

    # Constructor
    def __init__(self):
        self.head = None

    # Insert node at end
    def append(self, data):

        # Create new node
        new_node = Node(data)

        # Empty list case
        if not self.head:
            self.head = new_node
            return

        # Traverse to end
        current = self.head

        while current.next:
            current = current.next

        current.next = new_node

    # Delete node by value
    def delete_by_value(self, target):

        # Handle empty list
        if not self.head:
            return

        # Delete head node
        if self.head.data == target:
            self.head = self.head.next
            return

        # Traverse list
        current = self.head

        while current.next:

            # Target found
            if current.next.data == target:

                # Bypass target node
                current.next = current.next.next
                return

            current = current.next

    # Print linked list
    def display(self):

        current = self.head

        while current:
            print(current.data, end=" -> ")
            current = current.next

        print("None")


# Create linked list
linked_list = LinkedList()

# Insert values
linked_list.append(10)
linked_list.append(20)
linked_list.append(30)

# Delete value
linked_list.delete_by_value(20)

# Print updated list
linked_list.display()
````

**Answer:**  
Traverse the list while checking `current.next.data`. When the target is found, reconnect the pointers to bypass the target node.

**Explanation:**  
In a Singly Linked List, nodes only know about the next node.

To safely remove a node, the algorithm must stop one position before the target node so the chain can be rewired correctly.

The deletion process works by redirecting:

```python
current.next = current.next.next
```

This completely removes the target node from the chain.

**Time Complexity:**  
O(n)

**Space Complexity:**  
O(1)

##### **Question 1: Reverse a Singly Linked List**
**Problem Statement:** Assume you are provided the absolute Head pointer of a Singly Linked List. Reconstruct the entire structure so that all pointers violently change direction, completely reversing the list sequentially. Output the new Head pointer.
**Example Input:** Head targeting [one -> two -> three -> null]
**Example Output:** Head targeting [three -> two -> one -> null]

````python
# Define the function accepting the head node of the list securely
def reverse_list(head):
    # Initialize a tracking variable to store the previous node safely, defaulting to None
    previous = None
    # Initialize a temporary pointer to actively traverse the chain, starting at the head
    current = head

    # Loop aggressively while the current pointer holds a valid object
    while current is not None:
        # CRITICAL: Temporarily cache the next node before we violently overwrite the pointer
        next_temp = current.next

        # Execute the reversal: Force the current node to point backward explicitly
        current.next = previous

        # Shift the previous pointer forward to lock onto the current node
        previous = current
        # Shift the current pointer forward to securely target the cached next node
        current = next_temp

    # The loop terminates when current is null, meaning previous successfully holds the new Head
    return previous
````

**Time Complexity:** O(N) evaluation time, as the pointer must traverse every single node exactly once from the head to the trailing null value.

**Space Complexity:** O(1) allocation bounds. The function only instantiates primitive tracking variables internally, completely ignoring the creation of new Node objects.

##### **Question 2: Find Middle Node**
**Problem Statement:** Extract the mathematically central node of a given Singly Linked List. If the list possesses an explicitly even number of nodes, safely output the second middle node.
**Example Input:** Head targeting [one -> two -> three -> four -> five]
**Example Output:** Node(three)

````python
# Define the evaluation function utilizing the Fast and Slow pointer architectural pattern
def find_middle_node(head):
    # Setup the slow runner tracking pointer anchored exactly at the head
    slow = head
    # Setup the fast runner tracking pointer anchored identically at the head
    fast = head

    # Loop relentlessly while the fast pointer AND the node directly ahead of it securely exist
    while fast is not None and fast.next is not None:
        # Propel the slow runner forward strictly by one single leap
        slow = slow.next
        # Propel the fast runner forward aggressively by two explicit leaps
        fast = fast.next.next

    # When the fast runner violently hits the absolute edge of the list, the slow runner is perfectly centralized
    return slow
````

**Time Complexity:** O(N) bounds. The while loop evaluates roughly half the nodes due to the double-speed fast pointer.

**Space Complexity:** O(1) processing footprint, maintaining only two isolated tracking references.

##### **Question 3: Merge Two Sorted Lists**
**Problem Statement:** Assume you possess two strictly ascending Singly Linked Lists. Weave them together securely to form one unified, perfectly sorted master list. Synthesize the new list by physically splicing the original nodes.
**Example Input:** List A [one -> three], List B [two -> four]
**Example Output:** [one -> two -> three -> four]

````python
# Define a blank dummy class representation purely to simplify referencing
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

# Define the synthesis engine taking both list heads
def merge_two_lists(list1, list2):
    # Instantiate a dummy node cleanly to avoid dealing with empty head edge cases violently
    dummy = ListNode()
    # Initialize a trailing pointer anchored strictly to the dummy node
    tail = dummy

    # Loop dynamically while both primary lists hold viable evaluation nodes
    while list1 is not None and list2 is not None:
        # Compare the numerical values contained securely within the leading nodes
        if list1.val < list2.val:
            # Assign the tail's next pointer strictly to the smaller node from list one
            tail.next = list1
            # Advance the list one pointer securely to the next evaluation target
            list1 = list1.next
        else:
            # Assign the tail's next pointer strictly to the smaller node from list two
            tail.next = list2
            # Advance the list two pointer securely to the next evaluation target
            list2 = list2.next

    # Shift the tracking tail pointer aggressively forward to target the newly appended node
        tail = tail.next

    # Once the loop terminates, one list is entirely exhausted. Safely append the remainder of the surviving list.
    if list1 is not None:
        # Attach the surviving nodes seamlessly
        tail.next = list1
    elif list2 is not None:
        # Attach the surviving nodes seamlessly
        tail.next = list2

    # Output the unified list by explicitly bypassing the initial dummy node placeholder
    return dummy.next
````

**Time Complexity:** O(N + M) scaling logically where N and M map directly to the total nodes in each respective list.

**Space Complexity:** O(1) allocation because the logic physically stitches existing nodes rather than spawning duplicates.

##### **Question 4: Remove Nth Node From End**
**Problem Statement:** Assume a Singly Linked List and a positive integer K. Remove the mathematically precise K-th node from the absolute end of the list securely, and return the modified head pointer.
**Example Input:** [one -> two -> three -> four -> five], K = two
**Example Output:** [one -> two -> three -> five]

````python
# Define the targeted deletion function accepting the head and the trailing index
def remove_nth_from_end(head, n):
    # Generate a dummy node securely to handle complex deletions targeting the absolute head node violently
    dummy = ListNode(0, head)
    # Instantiate the left tracking pointer securely on the dummy
    left = dummy
    # Instantiate the right tracking pointer explicitly on the initial head
    right = head

    # Loop relentlessly to forcefully create a physical gap of exactly 'n' nodes between the pointers
    while n > 0:
        # Advance the right pointer forward strictly
        right = right.next
        # Decrement the loop counter to close the condition securely
        n -= 1

    # Loop steadily moving both pointers identically until the right pointer hits the null boundary
    while right is not None:
        # Advance the left pointer securely
        left = left.next
        # Advance the right pointer securely
        right = right.next

    # The left pointer now rests absolutely exactly one step behind the doomed target node
    # Brutally severe the target node from the logical sequence by leaping its next pointer
    left.next = left.next.next

    # Definitively return the modified sequence structure, bypassing the dummy securely
    return dummy.next
````

**Time Complexity:** O(N) evaluation bounds. The algorithm accomplishes the deletion utilizing a highly optimized single pass.

**Space Complexity:** O(1) processing footprint isolated to primitive pointer tracking exclusively.

##### **Question 5: Linked List Cycle Detection**
**Problem Statement:** Given the absolute head of a Linked List, explicitly determine if a chaotic infinite loop exists internally. A cycle occurs if any node mathematically references a node already visited previously.
**Example Input:** Head targets [one -> two -> three -> points back to two]
**Example Output:** True

````python
# Define the cycle verification function deploying Floyd's Tortoise and Hare paradigm
def has_cycle(head):
    # Instantiate the slow runner tracking pointer anchored exactly at the head
    slow = head
    # Instantiate the fast runner tracking pointer anchored identically at the head
    fast = head

    # Iterate persistently while the fast pointer and its subsequent target safely exist
    while fast is not None and fast.next is not None:
        # Propel the slow runner forward linearly by one discrete step
        slow = slow.next
        # Propel the fast runner forward exponentially by two distinct steps
        fast = fast.next.next

        # Analyze structurally if the fast runner mathematically lapped and collided with the slow runner
        if slow == fast:
            # Violent collision explicitly verifies the presence of an infinite closed track
            return True

    # If the fast pointer hits the null boundary entirely, the track is absolutely straight and safe
    return False
````

**Time Complexity:** O(N) constraints. The mathematical properties dictate the fast runner will flawlessly intersect the slow runner within a linear bound if a cycle breathes.

**Space Complexity:** O(1) processing overhead relying natively upon primitive locators alone.

##### **Question 6: Reorder List**
**Problem Statement:** Assume a Singly Linked List formatted as L0 → L1 → … → Ln-1 → Ln. Architect an algorithm that forcefully mutates the structure strictly in-place to follow the complex weaving format: L0 → Ln → L1 → Ln-1 → L2 → Ln-2.
**Example Input:** [one, two, three, four]
**Example Output:** [one, four, two, three]

````python
# Define the heavy list reconstruction architecture
def reorder_list(head):
    # Neutralize catastrophic edge cases immediately to prevent execution failures
    if not head or not head.next:
        # Abort cleanly
        return

    # Phase One: Uncover the exact mathematical center of the sequence securely
    slow, fast = head, head
    while fast and fast.next:
        # Step slow
        slow = slow.next
        # Leap fast
        fast = fast.next.next

    # Phase Two: Violently reverse the entire secondary trailing half of the list
    prev, curr = None, slow.next
    # Sever the primary half explicitly from the secondary half to prevent loop pollution
    slow.next = None

    while curr:
        # Cache next
        next_temp = curr.next
        # Invert pointer
        curr.next = prev
        # Shift previous
        prev = curr
        # Shift current
        curr = next_temp

    # Phase Three: Mathematically weave the primary half directly into the inverted secondary half
    first, second = head, prev
    while second:
        # Cache the next leading evaluation targets safely
        tmp1, tmp2 = first.next, second.next

        # Intersect the primary node aggressively to the secondary node
        first.next = second
        # Intersect the secondary node aggressively back to the next primary node
        second.next = tmp1

        # Shift the operational pointers deeply into the structure for the subsequent loop
        first, second = tmp1, tmp2
````

# Summary

## What did we learn
We learned that Linked Lists dismantle the rigid limitations of contiguous memory arrays by relying aggressively on dynamically allocated nodes and explicit referencing pointers. We established the architectural foundations of Singly Linked Lists, validating how they facilitate flawless O(1) insertions at the Head. We subsequently escalated our understanding to Doubly Linked Lists, proving how the addition of a previous tracking pointer unlocks omnidirectional traversal and drastically speeds up localized deletions at the cost of heavier RAM constraints.

## Best Practices & Tips
Always aggressively draw physical diagrams before attempting to write complex pointer manipulation code. Tracking `node.next.next` mentally almost universally leads to disastrous orphaned memory leaks.
Never lose the Head pointer. If you overwrite the overarching Head pointer blindly without securely storing a temporary reference to the original first node, the entire data structure becomes fundamentally unrecoverable garbage.
Deploy a dedicated Tail pointer whenever possible. It utterly eliminates the O(N) traversal penalty historically associated with appending data to the tail boundary.

## Shortcomings
Linked Lists suffer catastrophic performance drops when random access is heavily demanded. While an array returns `arr[500]` in constant time, a Linked List forces the processor to execute 500 individual sequential pointer hops simply to read the value. Furthermore, the objects lack cache locality; because nodes are scattered radically across the system heap, the CPU cannot aggressively cache the dataset, triggering continuous micro-delays.
