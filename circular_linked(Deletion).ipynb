# 1. Circular Linked List — Basic Structure

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Create nodes
a = Node(10)
b = Node(20)
c = Node(30)
d = Node(40)
e = Node(50)

# Connect nodes
a.next = b
b.next = c
c.next = d
d.next = e
e.next = a

# Head
head = a
```

# 2. Traversal Function

```python
def display(head):

    if head is None:
        print("List is empty")
        return

    curr = head

    while True:
        print(curr.data, end=" -> ")
        curr = curr.next

        if curr == head:
            break

    print("(back to head)")
```

# 3. Delete From Beginning

```python
def delete_beginning(head):

    if head is None:
        print("List is empty")
        return None

    # Only one node
    if head.next == head:
        return None

    # Find last node
    last = head

    while last.next != head:
        last = last.next

    # Move head
    head = head.next

    # Connect last node to new head
    last.next = head

    return head
```

# 4. Delete From End

```python
def delete_end(head):

    if head is None:
        print("List is empty")
        return None

    # Only one node
    if head.next == head:
        return None

    curr = head

    # Find second-last node
    while curr.next.next != head:
        curr = curr.next

    # Remove last node
    curr.next = head

    return head
```

# 5. Delete From Kth Position

```python
def delete_kth(head, k):

    if head is None:
        print("List is empty")
        return None

    if k <= 0:
        print("Invalid position")
        return head

    # Delete first node
    if k == 1:
        return delete_beginning(head)

    curr = head
    count = 1

    # Reach node before kth node
    while count < k - 1 and curr.next != head:
        curr = curr.next
        count += 1

    # Position does not exist
    if curr.next == head:
        print("Position does not exist")
        return head

    # Delete kth node
    curr.next = curr.next.next

    return head
```

```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:

    def __init__(self):
        self.head = None

    # Insert at end
    def insert_end(self, data):

        new_node = Node(data)

        # Empty list
        if self.head is None:
            self.head = new_node
            new_node.next = self.head
            return

        curr = self.head

        while curr.next != self.head:
            curr = curr.next

        curr.next = new_node
        new_node.next = self.head

    # Display
    def display(self):

        if self.head is None:
            print("List is empty")
            return

        curr = self.head

        while True:

            print(curr.data, end=" -> ")

            curr = curr.next

            if curr == self.head:
                break

        print("(back to head)")

    # Delete from beginning
    def delete_beginning(self):

        if self.head is None:
            print("List is empty")
            return

        # Only one node
        if self.head.next == self.head:
            self.head = None
            return

        last = self.head

        while last.next != self.head:
            last = last.next

        self.head = self.head.next

        last.next = self.head

    # Delete from end
    def delete_end(self):

        if self.head is None:
            print("List is empty")
            return

        # Only one node
        if self.head.next == self.head:
            self.head = None
            return

        curr = self.head

        while curr.next.next != self.head:
            curr = curr.next

        curr.next = self.head

    # Delete kth node
    def delete_kth(self, k):

        if self.head is None:
            print("List is empty")
            return

        if k <= 0:
            print("Invalid position")
            return

        # Delete first node
        if k == 1:
            self.delete_beginning()
            return

        curr = self.head
        count = 1

        while count < k - 1 and curr.next != self.head:

            curr = curr.next
            count += 1

        if curr.next == self.head:
            print("Position does not exist")
            return

        curr.next = curr.next.next

    # Delete by value
    def delete_value(self, value):

        if self.head is None:
            print("List is empty")
            return

        # Delete head
        if self.head.data == value:
            self.delete_beginning()
            return

        curr = self.head

        while curr.next != self.head:

            if curr.next.data == value:

                curr.next = curr.next.next

                return

            curr = curr.next

        print("Value not found")

    # Delete all
    def delete_all(self):

        self.head = None

        print("All nodes deleted")


# Create circular linked list

cll = CircularLinkedList()

cll.insert_end(10)
cll.insert_end(20)
cll.insert_end(30)
cll.insert_end(40)
cll.insert_end(50)


print("Original list:")
cll.display()


print("\nDelete from beginning:")
cll.delete_beginning()
cll.display()


print("\nDelete from end:")
cll.delete_end()
cll.display()


print("\nDelete 2nd position:")
cll.delete_kth(2)
cll.display()


print("\nDelete value 40:")
cll.delete_value(40)
cll.display()


print("\nDelete all nodes:")
cll.delete_all()
cll.display()
```

### Demonstrating Correct Usage of Standalone Deletion Functions

The standalone deletion functions (`delete_beginning`, `delete_end`, `delete_kth`) return the new `head` of the list. To ensure the list state is correctly updated, you need to re-assign the global `head` variable with the value returned by these functions.

```python
# Re-create the initial circular linked list for demonstration
# The Node class and initial 'head' (a,b,c,d,e) are defined in cell 'dnqmd8Svfdpg'

# Ensure the initial head and nodes are reset if you run this cell multiple times
a = Node(10)
b = Node(20)
c = Node(30)
d = Node(40)
e = Node(50)

a.next = b
b.next = c
c.next = d
d.next = e
e.next = a

head = a

print("Original list for standalone function demo:")
display(head)


print("\nDelete from beginning (correct usage):")
head = delete_beginning(head) # Re-assign head
display(head)


print("\nDelete from end (correct usage):")
head = delete_end(head) # Re-assign head
display(head)


print("\nDelete 2nd position (correct usage):")
head = delete_kth(head, 2) # Re-assign head
display(head)


print("\nAttempt to delete a non-existent position (correct usage):")
head = delete_kth(head, 5) # Re-assign head (will remain unchanged if position not found)
display(head)
```
